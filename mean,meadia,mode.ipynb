```python
import numpy as np

arr = np.random.randint(1,500,size = 10)
print(arr)

mean = np.mean(arr)
print("the mean of the give array is:",mean)
```

```python
arr = np.array([23,45,11,67,92,111,546,24,454,42,1,56,2])

print("the given array is\n",arr)

mdn = np.median(arr)
print("median of given arr is",mdn)
```

# **Standard Deviation**

```python
import numpy as np
import matplotlib.pyplot as plt
speed = np.random.randint(20,120, size = 7)

print(speed)

print("standard deviation is:")
std = np.std(speed)
print(std)
```

# **Variance**

```python
import numpy as np

speed = np.random.randint(20,120, size = 7)

print(speed)

print("standard deviation is:")
std = np.std(speed)
print(std)
var = np.var(std)
print("Variance is",var)
```

```python
arr = np.array([23,56,2,78,3,99,12,46,44])
std = np.std(arr)
var = np.var(arr)
print(std,var)
```

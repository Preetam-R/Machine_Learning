# Handle Null & Duplicates

```python
# Delete all the null values and duplicates
# fill the null values using Mean, Median, Mode.
# Mean (Average)
# Median (Middle after sorted) 4,5,2,7,4,8
# Mode (Most frequent values) [apple, orange, apple, banana,apple,banana,apple] [apple-mode]
```

```python
df.duplicated().sum()
```

```python
# To remove al the null values from the dataset
df.shape
# df.dropna(inplace=True)
df.drop_duplicates(inplace=True)
```

```python
df.columns
```

```python
# delete null values from cuisine and area

df.dropna(subset=['Cuisine', 'Area'], inplace=True)
```

```python
# df['Offer Name'] = df['Offer Name'].fillna(df['Offer Name'].mode()[0])
df['Number of Ratings'] = df['Number of Ratings'].fillna(df['Number of Ratings'].mode()[0])
```

```python
df.isnull().sum()
```

```python
df.head()
```

```python
# print the average number of offers running for veg and non-veg

df.groupby("Rating")['Number of Offers'].mean()
```

```python
df['Rating'] = df['Rating'].replace("NEW", 0)
df['Rating'] = df['Rating'].replace("--", 0)
```
